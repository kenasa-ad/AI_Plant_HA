```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 5, 10, 17, 26])

print("x =", x)
print("y =", y)
```

```python
x_squared = x ** 2

print("x² =", x_squared)
```

```python
plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Our Data")
plt.show()
```
